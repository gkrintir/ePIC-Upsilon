# Final-selection 2D efficiencies

Four separate figures for one selected sample. The axes use truth coordinates and the color shows the efficiency after **all final performance cuts**. Default is the new-ID 1S sample at 9×275 GeV. This notebook reads the original ROOT input and writes a fresh run. It does not rerun the cross-section extraction or modify existing runs.

The numerical CSV retains counts and 68.27% Wilson bounds. Gray cells have no eligible MC rows. Populated cells use the efficiency color scale without hatching. These are MC efficiency maps, not projected two-dimensional cross-section bins.

New runs default to **proton tagged**, requiring `reco_topology` 1 or 2 and positive finite `reco_tagged_proton_p`. The generated truth denominator is unchanged. Set `PROTON_SELECTION="inclusive"` only to reproduce the previous category. Before/after counts are raw MC yields, with no luminosity scaling. Use the new run ID so previous outputs remain preserved.


**To restyle an existing completed run**, open `05_final_efficiency_2d_display.ipynb` and Run All there. It shows each map separately with PNG/PDF links and does not rerun the ROOT calculation. This generation notebook still requires a fresh RUN_ID when executing its run cell.


This separate proton-tagged copy writes only to `Ypsilon-SWAN-ProtonTagged` by default. The earlier no-tag notebooks and results are preserved. The proton requirement is implemented and tested in this code; old figures are not regenerated until you execute a fresh run. These efficiencies use unweighted MC counts and finite-MC Wilson intervals; no luminosity, cross section, photon flux or weight branch enters.


```python
from pathlib import Path
import json
import subprocess
import sys
from IPython.display import display, Image

PACKAGE = Path.cwd().resolve()
if not (PACKAGE / "run_final_efficiency_2d.py").is_file():
    raise FileNotFoundError("Open this notebook from its Ypsilon_Efficiency_Diagnostics_ProtonTagged folder")
FRAMEWORK_ROOT = Path("/eos/user/g/gkrintir/Ypsilon-SWAN-ProtonTagged")
CONFIG = PACKAGE / "config_9x275.json"
SAMPLE = "1s_newid"
RUN_ID = "efficiency-2d-9x275-proton-v3"
HASH_INPUTS = True
PROTON_SELECTION = "required"  # inclusive reproduces the old category
```

Available samples are `1s_newid`, `1s_standard`, `2s_standard`, and `3s_standard`. Each run makes exactly four figures for the chosen sample. Use a fresh run ID if you change the sample or settings. Higher states retain their configured Q² maximum of 0.01 GeV².

```python
command = [sys.executable, str(PACKAGE / "run_final_efficiency_2d.py"),
           "validate", "--framework-root", str(FRAMEWORK_ROOT),
           "--config", str(CONFIG), "--sample", SAMPLE, "--proton-selection", PROTON_SELECTION]
subprocess.run(command, check=True)
```

```python
command[2] = "run"
run_command = command + ["--run-id", RUN_ID]
if HASH_INPUTS:
    run_command += ["--hash-inputs"]
subprocess.run(run_command, check=True)
```

Display the four standalone figures. When reopening an existing completed run, skip the run cell and execute the following cell directly.
